```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# 1. Dataset dari topik lpdp
corpus = [
    "Sudarto menegaskan alumni LPDP yang tidak patuh akan dikenakan sanksi sosial berupa publikasi nama di website resmi.",  # Doc 0 — sanksi
    "Sudarto menekankan dana yang digunakan membiayai pendidikan alumni merupakan uang rakyat yang berasal dari pajak.",      # Doc 1 — kebijakan
    "Penerima beasiswa LPDP wajib kembali ke Indonesia, alumni LPDP harus mengabdi kepada negara setelah menyelesaikan studi." # Doc 2 — kewajiban
]

# 2. Representasi Vektor (Hanya fokus pada kata 'alumni' dan 'lpdp')
vectorizer = CountVectorizer(vocabulary=['alumni', 'lpdp'])
X = vectorizer.fit_transform(corpus)

# 3. Hitung Cosine Similarity
# Membandingkan semua dokumen dengan Dokumen 0 (1 sanksi)
cos_sim = cosine_similarity(X[0:1], X)

# Tampilkan Hasil
df = pd.DataFrame(X.toarray(), columns=['alumni', 'lpdp'])
df['Cosine Sim to Doc 0'] = cos_sim.flatten()
print(df)
```

**CountVectorizer:** Mengubah teks menjadi angka (matriks frekuensi kata). Di sini kita membatasi fitur hanya pada kata "alumni" dan "lpdp" agar sesuai dengan grafik di slide Stanford.

**Cosine Similarity vs Distance:** Fungsi ini menghasilkan nilai 0 sampai 1. Semakin mendekati 1, semakin mirip kedua dokumen tersebut.

**Analisis Hasil:** Bahwa Dokumen 2 (kewajiban) memiliki nilai similarity yang tinggi terhadap Dokumen 0, membuktikan bahwa meskipun jumlah katanya jauh lebih banyak, "arah" topiknya tetap serupa.

```python
import re

# Derive documents dari corpus (hapus tanda baca & lowercase)
documents = [re.sub(r'[^\w\s]', '', doc.lower()) for doc in corpus]

# Tampilkan
for i, doc in enumerate(documents):
    print(f'Doc {i}: {doc}')
```

```python
import pandas as pd
from collections import Counter
tf = pd.DataFrame(
[Counter(doc.split()) for doc in documents],
).fillna(0)
tf
```

```python
import numpy as np
def length(v):
    return np.sqrt((v ** 2).sum())
def cos_dist(v, w):
    return 1 - (v * w).sum() / (length(v) * length(w))
cos_dist(tf.loc[0], tf.loc[1]), cos_dist(tf.loc[0], tf.loc[2])
```
